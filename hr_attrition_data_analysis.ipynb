```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\Dell\OneDrive\Documents\Project\Project 3\HR_Employee_Attrition_Data.csv")
```

```python
# ========================================================
# Task: Dataset Health Check & Summary Statistics
# Purpose: Inspect dimensions, data types, missing values,
#          and statistical summary of numerical features.
# ========================================================

print("---Dataset Shape (Rows, Columns) ---")
print(df.shape)
print("\n" + "="*40 + "\n")
```

```python
print("--- Dataset Infor & Data Types ---")
print(df.info())
print("\n" + "="*40 + "\n")
```

```python
print("--- Missing Values Count ---")
print(df.isnull().sum())
print("\n" + "="*40 + "\n")
```

```python
print("--- Statistical Summary of Numerical Columns ---")
print(df.describe().T)
```

```python
# ========================================================
# Task: Visualizing Monthly Income Distribution by Attrition
# Purpose: Compare income levels of employees who left vs stayed.
# ========================================================
```

```python
plt.figure(figsize = (8,5))
sns.boxplot(x = 'Attrition', y = 'MonthlyIncome', hue= 'Attrition', data=df, palette = 'Set2', legend=False)

#Graph title

plt.title('Monthly Income Distribution by Employee Attrition', fontsize = 14, fontweight = 'bold')
plt.xlabel('Attrition Status',fontsize =12)
plt.ylabel('Monthly Income (₹)', fontsize=12)

#Graph Display
plt.show()
```
